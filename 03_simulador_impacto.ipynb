# MOTOR 3: SIMULADOR DE IMPACTO NO TEMPO (VLT vs ÔNIBUS)
# Objetivo: Quantificar a economia de tempo para o trabalhador pessoense

```python
# 1. Definindo as Variáveis Iniciais (O Cenário)
distancia_rota_km = 12.5  # Ex: Distância média de Mangabeira até o Centro (Varadouro)
velocidade_onibus_kmh = 14.0 # Vel. média de um ônibus em JP no pico (trânsito + paradas)
velocidade_vlt_kmh = 35.0    # Vel. média do VLT em via segregada (sem trânsito)

# 2. O Motor Matemático (Física: Tempo = Distância / Velocidade)
# Convertendo o resultado (horas) direto para minutos multiplicando por 60
tempo_onibus_minutos = (distancia_rota_km / velocidade_onibus_kmh) * 60
tempo_vlt_minutos = (distancia_rota_km / velocidade_vlt_kmh) * 60

# 3. Calculando o Impacto na Vida (Economia de tempo)
economia_por_viagem = tempo_onibus_minutos - tempo_vlt_minutos
economia_diaria = economia_por_viagem * 2  # Ida e Volta

# Projeção anual considerando 22 dias úteis por mês em 12 meses (264 dias úteis)
economia_anual_minutos = economia_diaria * 264
economia_anual_horas = economia_anual_minutos / 60
economia_anual_dias = economia_anual_horas / 24 # Quantos dias inteiros a pessoa salva por ano

# 4. Gerando o Relatório Executivo (Output para a Reunião)
print("-" * 60)
print("📊 SIMULADOR DE IMPACTO DE MOBILIDADE: MANGABEIRA ➔ CENTRO")
print("-" * 60)
print(f"Distância do Trajeto: {distancia_rota_km} km\n")

print(f"🚌 Cenário Atual (Ônibus): O trajeto dura aprox. {tempo_onibus_minutos:.0f} minutos.")
print(f"🚆 Cenário Proposto (VLT): O trajeto durará aprox. {tempo_vlt_minutos:.0f} minutos.\n")

print(f"⏱️ Economia por Viagem (Ida): {economia_por_viagem:.0f} minutos.")
print(f"⏱️ Economia Diária (Ida/Volta): {economia_diaria:.0f} minutos.\n")

print("-" * 60)
print("🚀 IMPACTO DE VIDA DO TRABALHADOR EM 1 ANO:")
print(f"Substituindo o ônibus pelo VLT, o cidadão de João Pessoa recupera:")
print(f"• {economia_anual_horas:.0f} horas livres por ano.")
print(f"• Isso equivale a {economia_anual_dias:.1f} dias inteiros (24h) de vida devolvidos à população.")
print("-" * 60)
```
