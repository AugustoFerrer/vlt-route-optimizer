# MOTOR 2: MAPA DE CALOR E POLOS GERADORES DE VIAGENS
# Objetivo: Cruzar a rota otimizada com a densidade populacional para justificar o traçado

# Configurando ambiente de trabalho:

```python
import folium
from folium.plugins import HeatMap
import pandas as pd
import numpy as np
import io
```

# Lendo pontos geradores de viagem e mapeando sua demanda:

```python
pgv = pd.read_csv('dados/pgv.csv')

coordenadas_calor = []

for index, linha in pgv.iterrows():

    if linha['categoria'] == "Integracao / Transbordo":
        quantidade = 800  # Muito calor nos terminais
    elif linha['categoria'] == "Polo Educacional":
        quantidade = 400
    else:
        quantidade = 200
        
    # Espalhamento ao redor da coordenada real
    lats_espalhadas = np.random.normal(linha['latitude'], 0.001, quantidade)
    lons_espalhadas = np.random.normal(linha['longitude'], 0.001, quantidade)
    
    # Junta latitude e longitude e joga na lista de calor
    for lat, lon in zip(lats_espalhadas, lons_espalhadas):
        coordenadas_calor.append([lat, lon])
```

# Plotando o mapa com as zonas de calor (pgv):

```python
# 4. DESENHANDO O MAPA (A parte que você já domina do seu outro projeto!)
mapa_vlt = folium.Map(location=[-7.137, -34.86], zoom_start=13)

# Adiciona as manchas de calor no mapa
HeatMap(coordenadas_calor, radius=10, blur=8, max_zoom=1).add_to(mapa_vlt)

# Opcional: Adiciona os pinos reais para o candidato ver de onde a mancha saiu
for index, linha in pgv.iterrows():
    folium.Marker(
        [linha['latitude'], linha['longitude']], 
        popup=linha['nome'],
        icon=folium.Icon(color='black', icon='info-sign')
    ).add_to(mapa_vlt)

mapa_vlt
```
