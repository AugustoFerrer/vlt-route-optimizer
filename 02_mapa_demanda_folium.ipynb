```python
import folium
from folium.plugins import HeatMap
import pandas as pd
import numpy as np
import io


dados_epicentros_csv = """nome,latitude,longitude,categoria
"Terminal de Integracao do Varadouro (TIV)",-7.1195,-34.8875,"Integracao / Transbordo"
"Parque Solon de Lucena (Lagoa)",-7.1228,-34.8801,"Integracao / Transbordo"
"Terminal de Integracao do Valentina (TIVALE)",-7.1944,-34.8465,"Integracao / Transbordo"
"Campus I da UFPB (Castelo Branco)",-7.1396,-34.8449,"Polo Educacional"
"Centro Universitario UNIPE (Agua Fria)",-7.1517,-34.8569,"Polo Educacional"
"IFPB Campus Joao Pessoa (Jaguaribe)",-7.1345,-34.8752,"Polo Educacional"
"Manaira Shopping (Retao de Manaira)",-7.0991,-34.8453,"Varejo e Emprego"
"Mangabeira Shopping (Av. Hilton Souto Maior)",-7.1622,-34.8315,"Varejo e Emprego"
"Centro de Mangabeira (Av. Josefa Taveira)",-7.1678,-34.8340,"Varejo e Emprego"
"Centro Administrativo Municipal (CAM)",-7.1651,-34.8561,"Administrativo / Saude"
"Polo Medico da Torre (Complexo Hospitalar)",-7.1255,-34.8568,"Administrativo / Saude"
"""

# O Pandas lê esse texto como se fosse um arquivo Excel
df_epicentros = pd.read_csv(io.StringIO(dados_epicentros_csv))

# 2. LISTA VAZIA PARA GUARDAR O "CALOR"
coordenadas_calor = []

# 3. O LAÇO DE REPETIÇÃO INTELIGENTE (O seu motor gerador)
for index, linha in df_epicentros.iterrows():
    
    # Define quantas pessoas (pontos) vão aparecer baseado na categoria do local
    if linha['categoria'] == "Integracao / Transbordo":
        quantidade = 800  # Muito calor nos terminais
    elif linha['categoria'] == "Polo Educacional":
        quantidade = 400   # Calor forte nas faculdades
    else:
        quantidade = 200   # Calor médio no comércio/hospitais
        
    # Gera o espalhamento ao redor da coordenada real usando a matemática
    lats_espalhadas = np.random.normal(linha['latitude'], 0.001, quantidade)
    lons_espalhadas = np.random.normal(linha['longitude'], 0.001, quantidade)
    
    # Junta latitude e longitude e joga na nossa lista de calor
    for lat, lon in zip(lats_espalhadas, lons_espalhadas):
        coordenadas_calor.append([lat, lon])

# 4. DESENHANDO O MAPA (A parte que você já domina do seu outro projeto!)
mapa_vlt = folium.Map(location=[-7.137, -34.86], zoom_start=13)

# Adiciona as manchas de calor no mapa
HeatMap(coordenadas_calor, radius=10, blur=8, max_zoom=1).add_to(mapa_vlt)

# Opcional: Adiciona os pinos reais para o candidato ver de onde a mancha saiu
for index, linha in df_epicentros.iterrows():
    folium.Marker(
        [linha['latitude'], linha['longitude']], 
        popup=linha['nome'],
        icon=folium.Icon(color='black', icon='info-sign')
    ).add_to(mapa_vlt)

mapa_vlt
```
